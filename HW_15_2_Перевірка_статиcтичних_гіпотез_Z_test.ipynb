#1. Параметри задачі

```python
import numpy as np

data = data['data'].values

sample_mean = np.mean(data)
population_mean = 20
population_std = np.std(data, ddof=1)
sample_size = len(data)
alpha = 0.05

sample_mean, population_std, sample_size
```

#2. Нульова гіпотеза H₀: μ=20

   Альтернативна гіпотеза Hₐ: μ>20

   Тип тесту: правосторонній

# 3. Стандартна помилка SE

```python
SE = population_std / np.sqrt(sample_size)
SE
```

# 4. Z-статистика

```python
Z = (sample_mean - population_mean) / SE
Z
```

# 5. Критичне z-значення

```python
from scipy import stats

z_critical = stats.norm.ppf(1 - alpha)
z_critical
```

# 6. P-value

```python
p_value = 1 - stats.norm.cdf(Z)
p_value
```

#7. Висновок: За результатами аналізу Z-статистики та p-value при рівні значущості α = 0.05 ми не відхиляємо нульову гіпотезу H₀. Це означає, що середній чек не має статистично значущого зростання.

Виведіть p-значення та зробіть висновок, чи ми маємо достатньо доказів, аби стверджувати, що середній чек зріс.

Чи зійшлись значення z-статистику та р-значення в цьому завданні з попередніми обчисленнями?

```python
from statsmodels.stats.weightstats import ztest

z_stat, p_value_two_tailed = ztest(data, value=20)
p_value = 1 - p_value_two_tailed / 2

print("Z-statistic:", z_stat)
print("Two-tailed p-value:", p_value_two_tailed)
print("One-tailed p-value:", p_value)
```

Значення z-статистики та p-value збігаються з результатами, отриманими під час ручних обчислень.
