````python
import math


def calc_area(r):
    """Return the area of a circle."""
    return math.pi * r ** 2


def calc_circumference(r):
    """Return the circumference of a circle."""
    return 2 * math.pi * r


radius = 5

print(
    "Area:", calc_area(radius),
    "Circumference:", calc_circumference(radius)
)
````

````python
import pandas as pd
import numpy as np

np.random.seed(0)

products = ['apple', 'banana', 'cherry', 'dates', 'elderberry']
categories = ['fruit', 'fruit', 'fruit', 'fruit', 'fruit']

N_ROWS = 10**6

df = pd.DataFrame({
    'product_name': np.random.choice(products, N_ROWS),
    'product_category': np.random.choice(categories, N_ROWS),
    'product_price': np.random.uniform(1, 10, N_ROWS),
    'transaction_date': pd.date_range(start='1/1/2000', periods=N_ROWS, freq='h')
})

# Пам'ять ДО
memory_before = df.memory_usage(deep=True).sum()

# Перетворення в категорії
df['product_name'] = df['product_name'].astype('category')
df['product_category'] = df['product_category'].astype('category')

# Пам'ять ПІСЛЯ
memory_after = df.memory_usage(deep=True).sum()

# У скільки разів зменшилась пам'ять
reduction_ratio = memory_before / memory_after

print(f"Memory before: {memory_before / 1024**2:.2f} MB")
print(f"Memory after: {memory_after / 1024**2:.2f} MB")
print(f"Reduced in: {reduction_ratio:.2f} times")
````

- перетворила стовпці product_name та product_category у категоріальний тип (category)
- використала astype('category') для оптимізації пам'яті
- виміряла використання пам'яті до та після перетворення за допомогою memory_usage(deep=True)
- обчислила, у скільки разів зменшилось використання пам'яті
- досягла значного зменшення використання пам'яті (приблизно у 7 разів)
